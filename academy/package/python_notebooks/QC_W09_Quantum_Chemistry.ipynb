## 1. From molecular problem to qubit Hamiltonian

The electronic Hamiltonian is mapped to a finite qubit operator using a basis set, second quantisation and a mapping such as Jordan–Wigner. This notebook demonstrates the variational energy-estimation pattern on a toy operator.

```python
import numpy as np
from scipy.optimize import minimize

# Toy Hamiltonian H = -0.8 Z + 0.2 X
X = np.array([[0,1],[1,0]], dtype=complex)
Z = np.array([[1,0],[0,-1]], dtype=complex)
Hmol = -0.8*Z + 0.2*X
exact = np.linalg.eigvalsh(Hmol)[0]
print('Exact toy ground-state energy:', exact)
```

## 2. Variational ansatz and energy

```python
zero = np.array([1,0], dtype=complex)
def Ry(t):
    return np.array([[np.cos(t/2),-np.sin(t/2)],[np.sin(t/2),np.cos(t/2)]],dtype=complex)
def Rz(t):
    return np.diag([np.exp(-1j*t/2),np.exp(1j*t/2)])
def energy(p):
    state = Rz(p[1]) @ Ry(p[0]) @ zero
    return float(np.real(np.vdot(state, Hmol @ state)))
res = minimize(energy, [0.2,0.1], method='Nelder-Mead')
print('VQE estimate:', res.fun)
print('Exact:', exact)
assert abs(res.fun-exact) < 1e-5
```
