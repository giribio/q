## 1. Complex vectors and normalization

```python
import numpy as np

v = np.array([1+1j, 2-1j], dtype=complex)
norm = np.linalg.norm(v)
v_norm = v / norm
print('Norm before:', norm)
print('Normalized vector:', v_norm)
assert np.isclose(np.vdot(v_norm, v_norm), 1.0)
```

## 2. Inner products and orthogonality

```python
zero = np.array([1, 0], dtype=complex)
one = np.array([0, 1], dtype=complex)
print('Inner product <0|1>:', np.vdot(zero, one))
assert np.isclose(np.vdot(zero, one), 0)
```

## 3. Tensor products

```python
plus = (zero + one) / np.sqrt(2)
two_qubit = np.kron(plus, zero)
print('Tensor-product state:', two_qubit)
assert two_qubit.shape == (4,)
assert np.isclose(np.linalg.norm(two_qubit), 1.0)
```

## 4. Unitary validation

```python
X = np.array([[0,1],[1,0]], dtype=complex)
I = np.eye(2, dtype=complex)
print('X dagger X =\n', X.conj().T @ X)
assert np.allclose(X.conj().T @ X, I)
```

## 5. Born rule

```python
psi = np.array([1, 1j], dtype=complex) / np.sqrt(2)
probabilities = np.abs(psi)**2
print(probabilities)
assert np.isclose(probabilities.sum(), 1.0)
```
