## 1. Classical bit and probability distribution

```python
import numpy as np

classical_bit = 0
probability_distribution = np.array([0.5, 0.5])
print('Classical bit:', classical_bit)
print('Distribution over outcomes 0 and 1:', probability_distribution)
assert np.isclose(probability_distribution.sum(), 1.0)
```

## 2. Represent a pure qubit state

A normalized state is \(|\psi\rangle=\alpha|0\rangle+\beta|1\rangle\), with \(|\alpha|^2+|\beta|^2=1\).

```python
alpha = 1/np.sqrt(2)
beta = 1j/np.sqrt(2)
psi = np.array([alpha, beta], dtype=complex)
probs = np.abs(psi)**2
print('State:', psi)
print('Measurement probabilities:', probs)
assert np.isclose(np.linalg.norm(psi), 1.0)
assert np.allclose(probs, [0.5, 0.5])
```

## 3. Sample measurement outcomes

Sampling estimates probabilities; it does not reveal the full state vector.

```python
rng = np.random.default_rng(7)
shots = 1000
samples = rng.choice([0, 1], size=shots, p=probs)
counts = {int(k): int(v) for k, v in zip(*np.unique(samples, return_counts=True))}
print(counts)
print('Observed frequency of 0:', np.mean(samples == 0))
```
