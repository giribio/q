# Quantum search: Deutsch-Jozsa and Grover

**Quantum Computing Academy — Week 05**

Study oracle-based algorithms and amplitude amplification using a small, simulator-friendly search space.

> Run cells from top to bottom. This notebook is designed for a fresh Python kernel.

```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# Two search qubits; mark state |11> using a phase oracle.
qc = QuantumCircuit(2, 2)
qc.h([0,1])
qc.cz(0,1)  # phase flip on |11>
qc.h([0,1])
qc.x([0,1])
qc.h(1)
qc.cx(0,1)
qc.h(1)
qc.x([0,1])
qc.h([0,1])
qc.measure([0,1],[0,1])
print(qc.draw())
```

## Run Grover circuit

```python
sim = AerSimulator()
counts = sim.run(qc, shots=1024, seed_simulator=4).result().get_counts()
print(counts)
print('Most frequent bitstring:', max(counts, key=counts.get))
```

## Classical comparison

For N=4 items, classical unstructured search needs up to 4 queries in the worst case. Grover's algorithm uses O(√N) oracle queries asymptotically; small examples do not establish practical speedup.

```python
N = 4
classical_worst_case = N
grover_query_scale = np.sqrt(N)
print('Classical worst-case query count:', classical_worst_case)
print('Grover asymptotic scale (not exact finite-N count):', grover_query_scale)
```
