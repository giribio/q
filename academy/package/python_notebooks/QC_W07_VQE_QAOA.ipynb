Implement a minimal one-qubit VQE example and a small classical objective illustrating QAOA's optimisation loop.

> Run cells from top to bottom. This notebook is designed for a fresh Python kernel.

```python
import numpy as np
from scipy.optimize import minimize

# Hamiltonian H = Z; exact ground-state energy is -1.
Z = np.array([[1,0],[0,-1]], dtype=complex)
zero = np.array([1,0], dtype=complex)

def Ry(theta):
    return np.array([[np.cos(theta/2), -np.sin(theta/2)],
                     [np.sin(theta/2), np.cos(theta/2)]], dtype=complex)

def energy(params):
    state = Ry(params[0]) @ zero
    return float(np.real(np.vdot(state, Z @ state)))

result = minimize(energy, x0=[0.4], method='Nelder-Mead', options={'maxiter':500})
print(result.x, result.fun)
assert result.fun < -0.99
```

## Interpret VQE

The ansatz prepares a parameterized trial state. The quantum device estimates ⟨H⟩; a classical optimizer updates parameters to minimize the energy.

## Small Max-Cut objective (classical reference)

```python
import itertools
edges = [(0,1),(1,2),(2,3),(3,0),(0,2)]
n = 4

def cut_value(bits):
    return sum(bits[i] != bits[j] for i,j in edges)

candidates = list(itertools.product([0,1], repeat=n))
best = max(candidates, key=cut_value)
print('Best bit assignment:', best)
print('Maximum cut:', cut_value(best))
```
