# Quantum Fourier transform and phase estimation

**Quantum Computing Academy — Week 06**

Build the QFT matrix for small registers and estimate a known phase with a simple controlled-unitary circuit.

> Run cells from top to bottom. This notebook is designed for a fresh Python kernel.

```python
import numpy as np

def qft_matrix(N):
    j = np.arange(N)
    k = j.reshape((N,1))
    return np.exp(2j*np.pi*k*j/N) / np.sqrt(N)

F4 = qft_matrix(4)
print(F4)
assert np.allclose(F4.conj().T @ F4, np.eye(4))
```

## Apply QFT to a basis state

```python
basis_1 = np.array([0,1,0,0], dtype=complex)
transformed = F4 @ basis_1
print('Probabilities after QFT:', np.abs(transformed)**2)
assert np.allclose(np.abs(transformed)**2, np.ones(4)/4)
```

## Phase estimation intuition

For a unitary eigenstate satisfying U|u⟩=e^(2πiφ)|u⟩, phase estimation encodes an approximation to φ in a control register. The exercise below samples the binary approximation for a phase exactly representable with two bits.

```python
phi = 0.25
bits = 2
estimate = round(phi * (2**bits)) / (2**bits)
print('True phase:', phi)
print('Two-bit grid estimate:', estimate)
assert estimate == phi
```
