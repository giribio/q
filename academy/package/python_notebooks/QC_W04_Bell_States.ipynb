# Multi-qubit circuits and entanglement

**Quantum Computing Academy — Week 04**

Construct Bell states with Qiskit, inspect probabilities and understand correlated measurements. Install qiskit and qiskit-aer if needed.

> Run cells from top to bottom. This notebook is designed for a fresh Python kernel.

```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])
print(qc.draw())
```

## Run the circuit

```python
sim = AerSimulator()
result = sim.run(qc, shots=2048, seed_simulator=11).result()
counts = result.get_counts()
print(counts)
allowed = {'00', '11'}
assert set(counts).issubset(allowed)
assert abs(counts.get('00',0)/2048 - 0.5) < 0.1
```

## State-vector interpretation

Before measurement, the ideal state is \((|00\rangle+|11\rangle)/\sqrt{2}\). This is entangled: it cannot be written as a product of two single-qubit states.

```python
bell = np.array([1,0,0,1], dtype=complex)/np.sqrt(2)
print('State vector:', bell)
print('Probabilities:', np.abs(bell)**2)
assert np.allclose(np.abs(bell)**2, [0.5,0,0,0.5])
```
