Construct and diagonalise a small transverse-field Ising Hamiltonian and inspect its ground-state energy as a parameter changes.

> Run cells from top to bottom. This notebook is designed for a fresh Python kernel.

```python
import numpy as np
from numpy.linalg import eigvalsh
I = np.eye(2, dtype=complex)
X = np.array([[0,1],[1,0]], dtype=complex)
Z = np.array([[1,0],[0,-1]], dtype=complex)

# Two-spin transverse-field Ising model: H = -J Z0 Z1 - h(X0 + X1)
J = 1.0

def H_ising(h):
    return -J*np.kron(Z,Z) - h*(np.kron(X,I)+np.kron(I,X))

for h in [0.0, 0.5, 1.0, 1.5]:
    print(h, eigvalsh(H_ising(h))[0])
```

## Ground-state properties

```python
fields = np.linspace(0, 2, 41)
energies = np.array([eigvalsh(H_ising(h))[0] for h in fields])
print('Minimum energy:', energies.min())
print('Energy at h=0:', energies[0])
print('Energy at h=2:', energies[-1])
```

## Interpretation

This is a finite two-spin model, not a bulk-material calculation. Finite-size results illustrate Hamiltonian modelling but cannot alone establish a phase transition.
