# Quantum noise and hardware limitations

**Quantum Computing Academy — Week 11**

Use Qiskit Aer to compare ideal and noisy Bell-state measurement distributions. Noise is stochastic and hardware-specific.

> Run cells from top to bottom. This notebook is designed for a fresh Python kernel.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, depolarizing_error

qc = QuantumCircuit(2,2)
qc.h(0)
qc.cx(0,1)
qc.measure([0,1],[0,1])
ideal = AerSimulator()
ideal_counts = ideal.run(qc, shots=4000, seed_simulator=12).result().get_counts()
```

## Add a simple depolarising noise model

```python
noise = NoiseModel()
noise.add_all_qubit_quantum_error(depolarizing_error(0.03, 1), ['h'])
noise.add_all_qubit_quantum_error(depolarizing_error(0.05, 2), ['cx'])
noisy = AerSimulator(noise_model=noise)
noisy_counts = noisy.run(qc, shots=4000, seed_simulator=12).result().get_counts()
print('Ideal:', ideal_counts)
print('Noisy:', noisy_counts)
```

## Compare distributions

```python
def normalized(counts, keys=('00','01','10','11')):
    total = sum(counts.values())
    return {k: counts.get(k,0)/total for k in keys}
print('Ideal probabilities:', normalized(ideal_counts))
print('Noisy probabilities:', normalized(noisy_counts))
```
