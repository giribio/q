```python
import numpy as np
zero = np.array([1,0], dtype=complex)
I = np.eye(2, dtype=complex)
X = np.array([[0,1],[1,0]], dtype=complex)
Y = np.array([[0,-1j],[1j,0]], dtype=complex)
Z = np.array([[1,0],[0,-1]], dtype=complex)
H = np.array([[1,1],[1,-1]], dtype=complex)/np.sqrt(2)
S = np.diag([1,1j]).astype(complex)
T = np.diag([1,np.exp(1j*np.pi/4)]).astype(complex)
print('H|0> =', H @ zero)
assert np.allclose(H.conj().T @ H, I)
```

## Rotation gates

```python
def Rx(theta):
    return np.cos(theta/2)*I - 1j*np.sin(theta/2)*X

def Ry(theta):
    return np.array([[np.cos(theta/2), -np.sin(theta/2)],
                     [np.sin(theta/2), np.cos(theta/2)]], dtype=complex)

def Rz(theta):
    return np.diag([np.exp(-1j*theta/2), np.exp(1j*theta/2)])

for gate in [Rx(0.3), Ry(0.3), Rz(0.3)]:
    assert np.allclose(gate.conj().T @ gate, I)
print('Rotation gates are unitary.')
```

## Bloch-sphere coordinates (pure state)

```python
def bloch_coordinates(state):
    a, b = state
    return np.array([2*np.real(np.conj(a)*b), 2*np.imag(np.conj(a)*b), abs(a)**2-abs(b)**2])

state = H @ zero
print('Bloch vector for |+>:', bloch_coordinates(state))
```
