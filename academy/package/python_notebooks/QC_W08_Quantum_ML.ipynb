Build a simple quantum-kernel-inspired feature map and compare its Gram matrix with a classical baseline. Requires Qiskit.

> Run cells from top to bottom. This notebook is designed for a fresh Python kernel.

```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

def feature_state(x):
    qc = QuantumCircuit(1)
    qc.h(0)
    qc.rz(float(x), 0)
    return Statevector.from_instruction(qc)

def fidelity_kernel(x, y):
    sx, sy = feature_state(x), feature_state(y)
    return float(abs(np.vdot(sx.data, sy.data))**2)

X = np.linspace(-np.pi, np.pi, 12)
K = np.array([[fidelity_kernel(x,y) for y in X] for x in X])
print(K.round(3))
assert np.allclose(K, K.T)
assert np.all(np.linalg.eigvalsh(K) > -1e-8)
```

## Classical baseline

```python
K_classical = np.exp(-((X[:,None]-X[None,:])**2)/2)
print('Quantum kernel shape:', K.shape)
print('Classical RBF kernel shape:', K_classical.shape)
```

## Discussion

A quantum feature map is a data-encoding circuit. Kernel values are state fidelities in this example. This toy example does not establish a quantum advantage.
